### Question 20
Create a nested list and access elements inside it.

```python
nums = [[10, 20], [30, 40], [50, 60]]

print(nums[0])
print(nums[0][1])
print(nums[1][0])
print(nums[2][1])

for i in nums:
    for j in i:
        print(j)
```
