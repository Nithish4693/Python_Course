### Encapsulation
#### Encapsulation is the process of wrapping data and methods that operate on that data within a single unit, such as a class. It helps to protect the data from unauthorized access and modification. In Python, we can achieve encapsulation by using private and protected access modifiers.


```python
class BankAccount:
    def __init__(self,account_number,account_holder,balance):
        self.account_number = account_number
        self.account_holder = account_holder
        self.__balance = balance
        
    def deposit(self,amount):
        self.__balance += amount
        print(f"Deposited {amount}. New balance: {self.__balance}")
        
    def withdraw(self,amount):
        if amount > self.__balance:
            print("Insufficient funds")
        else:
            self.__balance -= amount
            print(f"Withdrew {amount}. New balance: {self.__balance}")
            
    def get_balance(self):
        return self.__balance
    

account = BankAccount("1234567890","John Doe",1000)
print(account.get_balance())
account.deposit(1500)
print(account.get_balance())
account.withdraw(200)
print(account.get_balance())

print(account.get_balance())

```
